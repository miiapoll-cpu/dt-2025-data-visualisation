```python
 import pandas as pd
```

```python
df = pd.read_csv("sports.csv")
df
```

```python
df.head()       #first rows
df.describe()   #statistics
```

```python
import matplotlib.pyplot as plt

df.plot(kind="bar", x="Name", y="Years of practice", legend=False)
plt.title("Years of practice per athlete")
plt.ylabel("Years")
plt.show()
```

```python
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.title("Age per athlete")
plt.ylabel("Age")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

#calculate quartiles and extremes

age_min = df["Age"].min()
age_q1 = df["Age"].quantile(0.25)
age_q3 = df["Age"].quantile(0.75)
age_max = df["Age"].max()

print("Age - Min:", age_min, "Q1", age_q1, "Q3", age_q3, "Max", age_max)

#Drw box plot
plt.figure(figsize=(8,5))
df[["Age"]].boxplot()
plt.show
```

```python


#calculate quartiles and extremes

yop_min = df["Years of practice"].min()
yop_q1 = df["Years of practice"].quantile(0.25)
yop_q3 = df["Years of practice"].quantile(0.75)
yop_max = df["Years of practice"].max()

print("Years of practice - Min:", yop_min, "Q1", yop_q1, "Q3", yop_q3, "Max", yop_max)

#Drw box plot
plt.figure(figsize=(8,5))
df[["Years of practice"]].boxplot()
plt.show
```

```python

#calculate quartiles and extremes

age_min = df["Age"].min()
age_q1 = df["Age"].quantile(0.25)
age_q3 = df["Age"].quantile(0.75)
age_max = df["Age"].max()

print("Age - Min:", age_min, "Q1", age_q1, "Q3", age_q3, "Max", age_max)

#Drw box plot
plt.figure(figsize=(8,5))
df[["Age", "Years of practice"]].boxplot() #boxploti loomiseks piisab siia "years of practice" i lisamisest
plt.show
```

```python
age_min = df["Age"].min()
age_q1 = df["Age"].quantile(0.25)
age_q3 = df["Age"].quantile(0.75)
age_max = df["Age"].max()

yop_min = df["Years of practice"].min()
yop_q1 = df["Years of practice"].quantile(0.25)
yop_q3 = df["Years of practice"].quantile(0.75)
yop_max = df["Years of practice"].max()

print("Years of practice - Min:", yop_min, "Q1", yop_q1, "Q3", yop_q3, "Max", yop_max)
print("Age - Min:", age_min, "Q1", age_q1, "Q3", age_q3, "Max", age_max)

#Drw box plot
plt.figure(figsize=(8,5)) #kui tahad suurust muuta, muuda figsize'i
df[["Age", "Years of practice"]].boxplot() #boxploti loomiseks piisab siia "years of practice" i lisamisest
plt.title("Box plot for age and years of practice") #saad panna pealkirja 
plt.show

```
