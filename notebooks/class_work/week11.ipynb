```python
sentence = "eestitalvonpime"

#eestitalvo n  p  i  m  e
#0123456789 10 11 12 13 14

print(sentence[0:3]) #start: end. extracted first 3 letters from the sentence. output is "tal"ArithmeticError
print(sentence[:3]) #can get the same output even without mentioning 0. output is "tal"

print(sentence[5:10])

#extract letters from 5 to end of the sentence
print(sentence[5:])
#extract last 4 letters
print(sentence[-4:])
#first letter of substring starting from index 5
print(sentence[5:][0]) #sentence[5:] will extract a substring starting from index 5
#[0] takes the first character of that string
#first 5 letter of substring from index 5
print(sentence[5:][:5])
#print entire string
print(sentence[:])
print(sentence)
#print every second character of the sentence
print(sentence[::2])
```

```python
sentence = "Kati ostis endale raamatu ja läks parki seda lugema"

words = sentence.split(" ")

print(words[0]) #print the first word

#print the first 3 words
print(words[0:3])
print(*words[0:3]) #without brackets and ülakoma

for w in words[0:3]: print(w) #gives the answer without brackets and prints each word in next line

#print the last 3 words
print("The last three words are: ", *words[-3:]) 
print(words[-3]) #prints the third word from the end

print("The  number of words is:", len(words)) 

#print the first letter of the third word
print(words[2][0])

#print the last letter of the last word
print(words[-1][-1])
```
